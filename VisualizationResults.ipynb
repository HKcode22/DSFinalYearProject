

## 1. Calibration Plot (One-vs-Rest)

**What it represents:**
This calibration plot shows how well the model's predicted probabilities match actual outcomes for 16 different classes for Funding Stage Prediction (labeled 0-15). Each line represents one class compared against all others (hence "one-vs-rest"). A perfectly calibrated model would follow the dashed diagonal line.

**Main characteristics:**
- The x-axis shows the mean predicted probability while the y-axis shows the actual observed fraction of positive outcomes
- Most classes show curves above the diagonal line, indicating that the model slightly underestimates probabilities (predicts lower probabilities than actual outcomes)
- Classes 0, 9, and 10 appear most closely calibrated (closest to diagonal)
- Classes 2, 12, and 13 show the most significant deviation from perfect calibration

**What the classes represent:**
These 16 classes (0-15) represent the different funding stages that startups can be classified into:
- Class 0: Pre-Seed
- Class 1: Seed
- Class 2: Angel
- Class 3: Series A
- Class 4: Series B
- Class 5: Series C
- Class 6: Series D
- Class 7: Series E
- Class 8: Series F
- Class 9: Series G
- Class 10: Series H
- Class 11: Venture - Series Unknown
- Class 12: Debt Financing
- Class 13: Undisclosed
- Class 14: Grant
- Class 15: Private Equity


## 2. Funding Amount vs. Employee Count by Stage

**What it represents:**
This scatter plot visualizes the relationship between a company's employee count (x-axis) and the funding amount in USD (y-axis), with points colored by funding stage.

**Main characteristics:**
- Both axes use logarithmic scales to accommodate the wide range of values
- Clear correlation between company size (employees) and funding amount
- Different funding stages show distinct patterns:
  - Pre-Seed and Seed companies (purple) cluster at lower employee counts and funding amounts
  - Series A (blue-purple) shows moderate funding and employee counts
  - Series B and C (blue) have higher values for both metrics
  - Series D+ (teal) reaches the highest funding amounts
- Some notable outliers exist, particularly companies with very high funding amounts (>$10B) relative to their employee count
- The general trend shows that as companies progress through funding stages, both employee count and funding amount tend to increase


## 3. Distribution of Success Scores

**What it represents:**
This histogram shows the distribution of success scores assigned to startups in the dataset, with threshold lines indicating risk categorization.

**Main characteristics:**
- Success scores range from approximately 20 to 75 (out of 100)
- The distribution is roughly normal with a slight right skew
- Three important thresholds are shown:
  - Red line (~40): High Failure Risk Threshold
  - Yellow line (~60): Success Threshold
  - Green line (~75): High Success Threshold
- The majority of startups fall in the middle range (40-60)
- The peak of the distribution is around 45-50, suggesting most companies have moderate success potential
- There are relatively few companies with very high success scores (>70)
- The blue curve shows the smoothed distribution (kernel density estimate)

**Effectiveness:**
This visualization effectively communicates the overall success landscape of the startup ecosystem, highlighting that most companies have moderate success potential while clearly delineating the risk categories through colored threshold lines.
